```python
LAST_NAME = "DUARIO"
SEED_NUM = 6
FAVORITE_ARTIST = "DINOROBONG TUBO"

fault_code = f"ERR-{LAST_NAME}-{SEED_NUM}-{len(FAVORITE_ARTIST)}"

def trace_fault(code, depth=1):
    print(f"Recursive Trace: Inspecting layer {depth} for {code}")
    
    # Base condition manually set to 2 to shorten the output for your paper
    if depth >= 2:
        print("[EXECUTION LOG] Base condition reached. Fault isolated.")
        return depth
        
    return trace_fault(code + f"-L{depth}", depth + 1)

print(f"Generated Fault Data: {fault_code}")
total_calls = trace_fault(fault_code)
print(f"Number of Recursive Calls: {total_calls}")
print(f"Final Output: Fault successfully traced through {total_calls} system layers.")
```
