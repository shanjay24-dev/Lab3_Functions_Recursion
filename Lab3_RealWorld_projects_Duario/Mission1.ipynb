```python
LAST_NAME = "DUARIO"
SEED_NUM = 6
FAVORITE_ARTIST = "DINOROBONG TUBO"

# 4. Decorator to record the diagnostic process
def diagnostic_logger(func):
    def wrapper(*args, **kwargs):
        print(f"[EXECUTION LOG] Initiating {func.__name__}...")
        result = func(*args, **kwargs)
        print("[EXECUTION LOG] Diagnostic complete.")
        return result
    return wrapper
```

```python
# 1. Generate student-specific equipment readings
def generate_readings():
    base_val = SEED_NUM * len(LAST_NAME)
    # Includes intentional strings to test invalid data handling
    return [base_val, base_val + len(FAVORITE_ARTIST), "SENSOR_FAIL", base_val * 2, "DISCONNECTED", 15]

# 3. Handle invalid input using exception handling
def validate_reading(reading):
    try:
        value = float(reading)
        return value
    except (ValueError, TypeError):
        return None
```

```python
# 2. Calculation and classification
@diagnostic_logger
def process_readings(readings):
    valid_readings = []
    invalid_count = 0
    
    for r in readings:
        val = validate_reading(r)
        if val is not None:
            valid_readings.append(val)
        else:
            invalid_count += 1
            
    average = sum(valid_readings) / len(valid_readings) if valid_readings else 0
    condition = "OPTIMAL" if average > 40 else "NEEDS MAINTENANCE"
    return valid_readings, invalid_count, round(average, 2), condition
```

```python
# 5. Display a summary based on the processed data
raw_data = generate_readings()
print(f"Generated Equipment Data: {raw_data}")

valid_data, invalid_errors, avg, status = process_readings(raw_data)

print(f"Validation Results: {len(valid_data)} Valid, {invalid_errors} Invalid")
print(f"Diagnostic Results: Average Output = {avg}")
print(f"Final Output: Equipment Condition is {status}")
```
